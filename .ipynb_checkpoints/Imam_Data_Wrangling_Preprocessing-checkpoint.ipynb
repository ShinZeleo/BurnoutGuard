### 3.5 Outlier Detection & Treatment

```python
# Identify numeric columnsnum_cols = df.select_dtypes(include=np.number).columns.tolist()print(f'Numeric columns ({len(num_cols)}): {num_cols}')# Detect outliers using IQR methodoutlier_summary = {}for col in num_cols:    Q1 = df[col].quantile(0.25)    Q3 = df[col].quantile(0.75)    IQR = Q3 - Q1    lower = Q1 - 1.5 * IQR    upper = Q3 + 1.5 * IQR    outliers = ((df[col] < lower) | (df[col] > upper)).sum()    if outliers > 0:        outlier_summary[col] = {            'count': outliers,            'pct': round(outliers/len(df)*100, 2),            'lower_bound': lower,            'upper_bound': upper,            'min': df[col].min(),            'max': df[col].max()        }if outlier_summary:    outlier_df = pd.DataFrame(outlier_summary).T    print('\n⚠️ Outliers detected (IQR method):')    print(outlier_df)else:    print('\n✅ No significant outliers detected.')
```

```python
# Visualize outliers with boxplotsfig, axes = plt.subplots(3, 4, figsize=(16, 10))axes = axes.flatten()for i, col in enumerate(num_cols):    if i < len(axes):        axes[i].boxplot(df[col], vert=True)        axes[i].set_title(col, fontsize=10)# Hide unused subplotsfor j in range(len(num_cols), len(axes)):    axes[j].set_visible(False)plt.suptitle('Boxplot — Outlier Detection (Numeric Features)', fontsize=14, fontweight='bold')plt.tight_layout()plt.show()
```

```python
# Treat outliers: Cap Stress_Score using IQR boundsif 'Stress_Score' in outlier_summary:    Q1 = df['Stress_Score'].quantile(0.25)    Q3 = df['Stress_Score'].quantile(0.75)    IQR = Q3 - Q1    lower = Q1 - 1.5 * IQR    upper = Q3 + 1.5 * IQR        before = len(df[(df['Stress_Score'] < lower) | (df['Stress_Score'] > upper)])    df['Stress_Score'] = df['Stress_Score'].clip(lower=lower, upper=upper)    after = len(df[(df['Stress_Score'] < lower) | (df['Stress_Score'] > upper)])        print(f'Stress_Score outliers capped: {before} -> {after}')    print(f'New range: [{df["Stress_Score"].min()}, {df["Stress_Score"].max()}]')else:    print('No outlier treatment needed for Stress_Score.')
```

---## 4. Feature Engineering

```python
# 4.1 Sleep Quality Ratio# Higher ratio = more sleep relative to study hours (better rest balance)df['Sleep_Quality_Ratio'] = df['Sleep_Hours'] / (df['Study_Hours'] + 1)  # +1 to avoid div by zeroprint('Sleep_Quality_Ratio stats:')print(df['Sleep_Quality_Ratio'].describe())
```

```python
# 4.2 Study vs Screen Ratio# Higher ratio = more study time relative to screen time (better focus)df['Study_Screen_Ratio'] = df['Study_Hours'] / (df['Screen_Time'] + 1)  # +1 to avoid div by zeroprint('Study_Screen_Ratio stats:')print(df['Study_Screen_Ratio'].describe())
```

```python
# 4.3 Stress Composite Feature# Weighted combination of stress-related indicators# Peer_Pressure + Anxiety_Level contribute positively to stress# Family_Support inversely contributes (higher support = less stress)df['Stress_Composite'] = (    df['Peer_Pressure'] * 0.3 +    df['Anxiety_Level'] * 0.4 +    (10 - df['Family_Support']) * 0.3  # Invert support)print('Stress_Composite stats:')print(df['Stress_Composite'].describe())
```

```python
# 4.4 Academic Load Index# Combined academic pressure metricdf['Academic_Load_Index'] = (    df['Study_Hours'] * 0.3 +    df['Assignment_Load'] * 0.35 +    df['Exam_Frequency'] * 0.35)print('Academic_Load_Index stats:')print(df['Academic_Load_Index'].describe())
```

```python
# 4.5 Digital Exposure# Total digital/screen engagementdf['Digital_Exposure'] = df['Social_Media_Use'] + df['Screen_Time']print('Digital_Exposure stats:')print(df['Digital_Exposure'].describe())
```

```python
# Summary of all features after engineeringprint(f'Total features: {df.shape[1]}')print(f'Total samples: {df.shape[0]}')print()print('All columns:')for i, col in enumerate(df.columns):    print(f'  {i:2d}. {col} ({df[col].dtype})')
```

```python
# Label Encoding for target variable (Stress_Level)le_target = LabelEncoder()df['Stress_Level_Encoded'] = le_target.fit_transform(df['Stress_Level'])print('Target Label Encoding:')for cls, idx in zip(le_target.classes_, le_target.transform(le_target.classes_)):    print(f'  {cls} -> {idx}')# Label Encoding for binary featuresle_gender = LabelEncoder()df['Gender_Encoded'] = le_gender.fit_transform(df['Gender'])print(f'\nGender: {dict(zip(le_gender.classes_, le_gender.transform(le_gender.classes_)))}')le_tuition = LabelEncoder()df['Tuition_Encoded'] = le_tuition.fit_transform(df['Tuition'])print(f'Tuition: {dict(zip(le_tuition.classes_, le_tuition.transform(le_tuition.classes_)))}')le_exercise = LabelEncoder()df['Physical_Exercise_Encoded'] = le_exercise.fit_transform(df['Physical_Exercise'])print(f'Physical_Exercise: {dict(zip(le_exercise.classes_, le_exercise.transform(le_exercise.classes_)))}')
```

```python
# One-Hot Encoding for multi-class categorical featuresdf_encoded = pd.get_dummies(df, columns=['Family_Income_Level', 'University_Type'],                             prefix=['Income', 'Uni'], drop_first=False, dtype=int)# Drop original categorical columns (already encoded)cols_to_drop = ['Gender', 'Tuition', 'Physical_Exercise', 'Stress_Level']df_encoded = df_encoded.drop(columns=cols_to_drop)print(f'Shape after encoding: {df_encoded.shape}')print(f'\nAll columns ({len(df_encoded.columns)}):')for i, col in enumerate(df_encoded.columns):    print(f'  {i:2d}. {col} ({df_encoded[col].dtype})')
```
